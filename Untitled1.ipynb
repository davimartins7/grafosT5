```python
# 1. Recriar o grafo e rodar o DSatur para o notebook "enxergar" as variáveis
data = open("dados/brasil.txt").read().strip().split('\n')
V = int(data[0])
E = int(data[1])
g = Graph(V)
for i in range(2, 2 + E):
    v, w = data[i].split()
    g.add_edge(v, w)

# Importar a função dsatur do seu arquivo ou definir aqui rapidamente
def get_colors(graph):
    n = graph.V
    colors = [-1] * n
    remaining = set(range(n))

    # Primeiro vértice (maior grau)
    start_v = max(range(n), key=lambda x: graph.degree(x))
    colors[start_v] = 0
    remaining.remove(start_v)

    while remaining:
        # DSatur: Escolher por saturação, depois por grau
        v = max(remaining, key=lambda x: (len({colors[nb] for nb in graph.adj[x] if colors[nb] != -1}), graph.degree(x)))
        used = {colors[nb] for nb in graph.adj[v] if colors[nb] != -1}
        c = 0
        while c in used: c += 1
        colors[v] = c
        remaining.remove(v)
    return colors

colors = get_colors(g)
states = ["AC", "AL", "AM", "AP", "BA", "CE", "DF", "ES", "GO", "MA", "MG", "MS", "MT", "PA", "PB", "PE", "PI", "PR", "RJ", "RN", "RO", "RR", "RS", "SC", "SE", "SP", "TO"]

# Coordenadas Geográficas Ajustadas para evitar sobreposição no Nordeste
pos = {
    "RR": (-60, 2),    "AP": (-51, 1),    "AM": (-63, -3),   "PA": (-52, -3),
    "MA": (-45, -5),   "CE": (-39, -5),
    "RN": (-35, -6),   # Afastado para a direita
    "PB": (-34, -7.5), # Afastado para a direita e baixo
    "PE": (-35, -8.5),   # Afastado para baixo
    "PI": (-42, -8),
    "AL": (-35, -10),  # Afastado para a direita
    "SE": (-37, -11.5), # Afastado para baixo e direita
    "AC": (-70, -9),   "RO": (-63, -11),  "TO": (-48, -10),  "BA": (-41, -13),
    "MT": (-55, -13),  "GO": (-49, -14),  "DF": (-47.9, -15.7), "MG": (-44, -19),
    "ES": (-40, -19),  "MS": (-54, -20),  "SP": (-48, -23),  "RJ": (-42, -22),
    "PR": (-51, -25),  "SC": (-50, -27),  "RS": (-53, -30)
}

plt.figure(figsize=(12, 14))

# Cores para o gráfico (ajustadas para os nomes que você escolheu)
mapa_cores = ["yellow", "cyan", "green", "red", "orange"]

# 1. Desenhar as conexões (fronteiras)
for v in range(g.V):
    x, y = pos[states[v]]
    for neighbor in g.adj[v]:
        nx, ny = pos[states[neighbor]]
        plt.plot([x, nx], [y, ny], color='black', linestyle='-', alpha=0.1, zorder=1)

# 2. Desenhar os estados com tamanho ligeiramente menor para não baterem
for v in range(g.V):
    x, y = pos[states[v]]
    plt.scatter(x, y, s=1100, c=mapa_cores[colors[v]], edgecolors='black', linewidth=1.2, zorder=2)
    plt.text(x, y, states[v], fontsize=9, ha='center', va='center', fontweight='bold', zorder=3)

plt.title("Grafo de Adjacência: Mapa do Brasil Colorido (DSatur)", fontsize=16, pad=20)
plt.axis('off')
plt.show()
```
